## A table with house prices in California (years ago) and correlated variables (features) 

```python
import os
import tarfile
import urllib.request

# Define function to download the data from the internet

DOWNLOAD_ROOT = "https://raw.githubusercontent.com/ageron/handson-ml2/master/"
HOUSING_PATH = os.path.join(".", "../datasets/housing")
HOUSING_URL = DOWNLOAD_ROOT + "datasets/housing/housing.tgz"

def fetch_housing_data(housing_url=HOUSING_URL, housing_path=HOUSING_PATH):
    if not os.path.isdir(housing_path):
        os.makedirs(housing_path)
    tgz_path = os.path.join(housing_path, "housing.tgz")
    urllib.request.urlretrieve(housing_url, tgz_path)
    housing_tgz = tarfile.open(tgz_path)
    housing_tgz.extractall(path=housing_path)
    housing_tgz.close()    
```

```python
fetch_housing_data()
```

```python
import pandas as pd

# Define function to load the data from a csv table

def load_housing_data(housing_path=HOUSING_PATH):
    csv_path = os.path.join(housing_path, "housing.csv")
    return pd.read_csv(csv_path)
```

## Show the features x1, x2, ..., xn. Note that they are not in the same scale, some of them are textual features, and others might be missing from the table. This means that any dataset requires preprocessing to normalize the data into a common scale, remove irrelevant information, complete or disregard incomplete information, convert text to numbers, etc., to make the dataset suitable and consistent for data analysis.

**Important:** A common mistake is to preprocess the entire dataset before splitting it into training, validation, and test sets. You should fit the preprocessing parameters (e.g., mean, standard deviation) on the training set only, and then apply them to the validation and test sets.

```python
housing = load_housing_data()
housing.head()
```

## Visualizing the raw data

Before any preprocessing, you can visualize the data in several different ways to better understand the problem. For instance, you can show the map of California color-coded by the prices of the houses.

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
mpl.rc('axes', labelsize=14)
mpl.rc('xtick', labelsize=12)
mpl.rc('ytick', labelsize=12)

# Where to save the figures
PROJECT_ROOT_DIR = "./housing"
IMAGES_PATH = os.path.join(PROJECT_ROOT_DIR, "images")
os.makedirs(IMAGES_PATH, exist_ok=True)

def save_fig(fig_id, tight_layout=True, fig_extension="png", resolution=300):
    path = os.path.join(IMAGES_PATH, fig_id + "." + fig_extension)
    print("Saving figure", fig_id)
    if tight_layout:
        plt.tight_layout()
    plt.savefig(path, format=fig_extension, dpi=resolution)
```

## Convert the dataset to numbers only and normalize it

Now we preprocess the data. For simplicity, we replace the categorical feature `ocean_proximity` with arbitrary numerical values and normalize the entire dataset. In practice, **one-hot encoding** is the standard approach for categorical features, and normalization parameters should be computed on the training set only (see note above).

```python
housing.ocean_proximity.replace(['<1H OCEAN','INLAND','NEAR OCEAN','NEAR BAY','ISLAND'], [0.4, 0.1, 0.8, 0.9, 0.7], inplace=True)
housing.iloc[:,0:-1] = housing.iloc[:,0:-1].apply(lambda x: (x-x.mean())/ x.std(), axis=0)
housing.head()
```

# Example of unstructured dataset (images)

The housing dataset above is **structured** (tabular): each sample is a row with named features. Now let's look at **unstructured** data — images — where each sample is a matrix of pixel values with spatial structure that cannot be captured by a simple feature table.

```python
from torchvision import datasets
import torch

data_folder = '../images' # This can be any folder you want to download FMNIST to
fmnist = datasets.FashionMNIST(data_folder, download=True, train=True) # download the training set
```

## Get the images and their classes (targets) as numpy arrays

```python
tr_images  = fmnist.data
tr_targets = fmnist.targets
```

```python
# Get a unique number for each class in a list, since the classes are names and not numbers. 
unique_values = tr_targets.unique()
print(f'tr_images & tr_targets:\n\tX - {tr_images.shape}\n\tY - {tr_targets.shape}\n\tY - Unique Values : {unique_values}')
print(f'TASK:\n\t{len(unique_values)} class Classification')
print(f'UNIQUE CLASSES:\n\t{fmnist.classes}') 
```

## Show examples of  images from each class

```python
import matplotlib.pyplot as plt
R, C = len(tr_targets.unique()), 10
fig, ax = plt.subplots(R, C, figsize=(10,10))
for label_class, plot_row in enumerate(ax):
    label_x_rows = np.where(tr_targets == label_class)[0]
    for plot_cell in plot_row:
        plot_cell.grid(False); plot_cell.axis('off')
        ix = np.random.choice(label_x_rows)
        x, y = tr_images[ix], tr_targets[ix]
        plot_cell.imshow(x, cmap='gray')
plt.tight_layout()
plt.show()
```

In the case of images, preprocessing might include rescaling, filtering, color conversions, normalization, etc. As discussed in the lecture, pretrained models on ImageNet expect specific normalization values per channel.
